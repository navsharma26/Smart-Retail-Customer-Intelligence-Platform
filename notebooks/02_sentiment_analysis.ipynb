# Sentiment Analysis Training & TF-IDF Vectorizer Pipeline
This notebook demonstrates building and training a TF-IDF + Logistic Regression Sentiment Analyzer on retail product reviews.

```python
import os
import pickle
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

def preprocess_text(text):
    text = text.lower().strip()
    text = re.sub(r'[^\w\s]', ' ', text)
    tokens = [w for w in text.split() if len(w) > 1]
    return ' '.join(tokens)

print('Text Preprocessor initialized.')
```
